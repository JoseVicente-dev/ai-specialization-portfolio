## Ejercicio 1
En un grupo de **100 estudiantes** de bachillerato que están cursando el último año:

| Dato | Valor |
|------|-------|
| Cursaron Matemáticas (M) | 42 |
| Cursaron Psicología (P) | 68 |
| Cursaron Historia (H) | 54 |
| Matemáticas e Historia (M ∩ H) | 22 |
| Matemáticas y Psicología (M ∩ P) | 25 |
| Historia pero **ni** Matemáticas **ni** Psicología (H ∩ M' ∩ P') | 7 |
| Las tres materias (M ∩ P ∩ H) | 10 |
| Ninguna de las tres | 8 |

**a)** ¿Probabilidad de que una persona elegida al azar esté inscrita en Psicología **y** curse las tres materias?  
**b)** ¿Probabilidad de que **no** esté inscrita en Psicología y esté cursando Historia **y** Matemáticas?

```python
# ── Datos del problema ─────────────────────────────────────────────────────────
n_total = 100

M   = 42   # |M|
P   = 68   # |P|
H   = 54   # |H|
MH  = 22   # |M ∩ H|
MP  = 25   # |M ∩ P|
MPH = 10   # |M ∩ P ∩ H|
H_solo = 7 # |H ∩ M' ∩ P'|
ninguna = 8

# ── Derivación de |P ∩ H| usando Inclusión-Exclusión ─────────────────────────
# |M ∪ P ∪ H| = n_total - ninguna
union = n_total - ninguna

# |M ∪ P ∪ H| = |M| + |P| + |H| - |M∩P| - |M∩H| - |P∩H| + |M∩P∩H|
# => |P∩H| = |M| + |P| + |H| - |M∩P| - |M∩H| + |M∩P∩H| - |M ∪ P ∪ H|
PH = M + P + H - MP - MH + MPH - union
print(f"|M ∪ P ∪ H| = {union}")
print(f"|P ∩ H|     = {PH}")

# Verificación: |H ∩ M' ∩ P'| = |H| - |M∩H| - |P∩H| + |M∩P∩H|
verificacion = H - MH - PH + MPH
print(f"\nVerificación |H ∩ M' ∩ P'| = {verificacion}  (debe ser {H_solo}) ✓" if verificacion == H_solo else f"Error: {verificacion} ≠ {H_solo}")
```

```python
# ── a) P(P ∩ M ∩ H) ────────────────────────────────────────────────────
# Estudiantes inscritos en Psicología Y que cursan las tres materias = M ∩ P ∩ H
prob_a = MPH / n_total
print(f"a) P(P ∩ M ∩ H) = {MPH}/{n_total} = {prob_a:.4f}")
print(f"   En fracción exacta: {Fraction(MPH, n_total)}")

# ── b) P(M ∩ H ∩ P') ───────────────────────────────────────────────────
# No inscrita en Psicología, cursando Historia Y Matemáticas
# |M ∩ H ∩ P'| = |M ∩ H| - |M ∩ H ∩ P|
MH_noP = MH - MPH
prob_b = MH_noP / n_total
print(f"\nb) |M ∩ H ∩ P'| = |M∩H| - |M∩H∩P| = {MH} - {MPH} = {MH_noP}")
print(f"   P(M ∩ H ∩ P') = {MH_noP}/{n_total} = {prob_b:.4f}")
print(f"   En fracción exacta: {Fraction(MH_noP, n_total)}")
```

## Ejercicio 2

Un fabricante de vacuna para la gripe tiene **tres departamentos** que inspeccionan los lotes de suero de forma **secuencial e independiente**, con las siguientes tasas de rechazo:

| Departamento | Tasa de rechazo |
|---|---|
| D1 | 0.10 |
| D2 | 0.08 |
| D3 | 0.12 |

**a)** ¿Probabilidad de que un lote sobreviva la primera inspección pero sea rechazado por el segundo departamento?  
**b)** ¿Probabilidad de que un lote sea rechazado por el tercer departamento? (es decir, pase D1 y D2, y sea rechazado en D3)

```python
# ── Datos ─────────────────────────────────────────────────────────────────────
p_rechazo  = np.array([0.10, 0.08, 0.12])   # tasas de rechazo por depto
p_aprueba  = 1 - p_rechazo                   # tasas de aprobación

print("Probabilidades de aprobación por departamento:")
for i, (pa, pr) in enumerate(zip(p_aprueba, p_rechazo), 1):
    print(f"  D{i}: P(aprueba) = {pa:.2f}  |  P(rechaza) = {pr:.2f}")

# ── Inciso a) P(pasa D1 ∩ rechazado D2) ──────────────────────────────────────
# Eventos independientes → se multiplican
prob_2a = p_aprueba[0] * p_rechazo[1]
print(f"\na) P(pasa D1 ∩ rechazado D2) = P(pasa D1) × P(rechaza D2)")
print(f"   = {p_aprueba[0]:.2f} × {p_rechazo[1]:.2f} = {prob_2a:.4f}")

# ── Inciso b) P(pasa D1 ∩ pasa D2 ∩ rechazado D3) ───────────────────────────
# El lote llega a D3 sólo si pasó D1 y D2
prob_2b = p_aprueba[0] * p_aprueba[1] * p_rechazo[2]
print(f"\nb) P(pasa D1 ∩ pasa D2 ∩ rechazado D3) = {p_aprueba[0]:.2f} × {p_aprueba[1]:.2f} × {p_rechazo[2]:.2f} = {prob_2b:.5f}")
```

## Ejercicio 3
- La probabilidad de que un doctor diagnostique **correctamente** una enfermedad es **0.7**.
- Dado que el diagnóstico es **incorrecto**, la probabilidad de que el paciente entable una demanda legal es **0.9**.

**¿Cuál es la probabilidad de que el doctor haga un diagnóstico incorrecto Y el paciente lo demande?**

```python
# ── Datos ─────────────────────────────────────────────────────────────────────
p_correcto          = 0.7
p_incorrecto        = 1 - p_correcto          # P(diagnóstico incorrecto)
p_demanda_dado_inc  = 0.9                     # P(demanda | incorrecto)

# ── Regla de la multiplicación ─────────────────────────────────────────────────
# P(incorrecto ∩ demanda) = P(incorrecto) × P(demanda | incorrecto)
p_inc_y_demanda = p_incorrecto * p_demanda_dado_inc

# ── Resumen en tabla pandas ────────────────────────────────────────────────────
tabla = pd.DataFrame({
    'Evento': [
        'P(diagnóstico correcto)',
        'P(diagnóstico incorrecto)',
        'P(demanda | diagnóstico incorrecto)',
        'P(incorrecto ∩ demanda)  ← RESPUESTA'
    ],
    'Valor': [
        p_correcto,
        p_incorrecto,
        p_demanda_dado_inc,
        p_inc_y_demanda
    ]
})
tabla['Fracción'] = tabla['Valor'].apply(lambda x: str(Fraction(x).limit_denominator(100)))
print(tabla.to_string(index=False))
print(f"\nResultado: P(incorrecto ∩ demanda) = {p_incorrecto:.1f} × {p_demanda_dado_inc:.1f} = {p_inc_y_demanda:.2f}")
```

## Ejercicio 4

Cuatro inspectores colocan la fecha de caducidad en los paquetes de película:

| Inspector | % de paquetes | Falla (no coloca fecha) |
|-----------|--------------|-------------------------|
| John      | 20%          | 1 de cada 200           |
| Tom       | 60%          | 1 de cada 100           |
| Jeff      | 15%          | 1 de cada 90            |
| Pat       | 5%           | 1 de cada 200           |

Si un paquete **no muestra la fecha de caducidad**, ¿cuál es la probabilidad de que haya sido inspeccionado por **John**?

```python
# ── Datos (probabilidades de cada inspector) ─────────────────────────
inspectores = ['John', 'Tom', 'Jeff', 'Pat']

p_prior = np.array([0.20, 0.60, 0.15, 0.05])   # P(inspector)
p_falla = np.array([1/200, 1/100, 1/90, 1/200]) # P(sin fecha | inspector)

# ── Teorema de la Probabilidad Total ─────────────────────────────────────────
# P(sin fecha) = Σ P(sin fecha | Di) × P(Di)
p_no_fecha = np.sum(p_falla * p_prior)
print(f"P(sin fecha) = {p_no_fecha:.8f}")

# ── Teorema de Bayes ──────────────────────────────────────────────────────────
# P(Di | sin fecha) = P(sin fecha | Di) × P(Di) / P(sin fecha)
p_posterior = (p_falla * p_prior) / p_no_fecha

# ── Tabla resumen ──────────────────────────────────────────────────
df_bayes = pd.DataFrame({
    'Inspector':            inspectores,
    'P(inspector)':         p_prior,
    'P(sin fecha|insp.)':   p_falla,
    'P(insp.)×P(sf|insp.)': p_falla * p_prior,
    'P(inspector|sin fecha)': p_posterior
})
df_bayes = df_bayes.set_index('Inspector')

# Fila de totales
totales = pd.DataFrame({
    'P(inspector)':         [p_prior.sum()],
    'P(sin fecha|insp.)':   ['—'],
    'P(insp.)×P(sf|insp.)': [p_no_fecha],
    'P(inspector|sin fecha)': [p_posterior.sum()]
}, index=['TOTAL'])

df_display = pd.concat([df_bayes, totales])
pd.set_option('display.float_format', '{:.6f}'.format)
display(df_display)

print(f"\nRespuesta: P(John | sin fecha) = {p_posterior[0]:.6f} ≈ {p_posterior[0]*100:.2f}%")
```
