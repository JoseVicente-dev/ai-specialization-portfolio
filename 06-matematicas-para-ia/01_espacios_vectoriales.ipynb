## <center><font color=#3FA1BF>Ejercicio 1 — Independencia Lineal</font></center>

Determine si los siguientes vectores son **linealmente independientes o dependientes**.

Para un conjunto $\{v_1, v_2, \ldots, v_n\}$, se resuelve el sistema homogéneo:
$$a_1 v_1 + a_2 v_2 + \cdots + a_n v_n = \mathbf{0}$$
- Si la **única solución** es $a_1 = a_2 = \cdots = a_n = 0$ (solución trivial) $\Rightarrow$ **INDEPENDIENTES**
- Si existe alguna **solución no trivial** $\Rightarrow$ **DEPENDIENTES**

**(a)** $\{(1,1);\;(4,4);\;(5,3)\}$

**(b)** $\{(3,5);\;(1,3);\;(0,0)\}$

**(c)** $\{(1,2,3,4);\;(1,0,0,0);\;(0,1,0,0)\}$

**(d)** $\{(1,1,1);\;(2,2,2);\;(0,0,1)\}$

**(e)** $\left\{\begin{pmatrix}1 & 2\\3 & 4\end{pmatrix},\;\begin{pmatrix}1 & 0\\0 & 1\end{pmatrix},\;\begin{pmatrix}1 & 0\\0 & 0\end{pmatrix}\right\}$ — matrices de $M_{2\times 2}$, tratadas como vectores en $\mathbb{R}^4$

```python
# Ejercicio 1 — Independencia lineal
# Para cada conjunto planteamos el sistema homogéneo a1*v1 + a2*v2 + ... = 0
# La matriz aumentada tiene los vectores como columnas y la última columna en ceros.

a1, a2, a3 = sympy.symbols('a1 a2 a3')

def clasificar(sol, variables):
    """Determina si el sistema homogéneo solo tiene solución trivial."""
    if sol is None:
        return "INDEPENDIENTES (solución trivial)"
    libres = set(variables) - set(sol.keys())
    if libres or any(v != 0 for v in sol.values()):
        return "DEPENDIENTES (existen soluciones no triviales)"
    return "INDEPENDIENTES (única solución es la trivial)"

# --- (a) {(1,1), (4,4), (5,3)} en R² ---
print("--- (a) {(1,1), (4,4), (5,3)} en R² ---")
A = sympy.Matrix([[1, 4, 5, 0],
                  [1, 4, 3, 0]])
sol = sympy.solve_linear_system(A, a1, a2, a3)
print("Solución sistema homogéneo:", sol)
print("→", clasificar(sol, [a1, a2, a3]))

# --- (b) {(3,5), (1,3), (0,0)} en R² ---
print("\n--- (b) {(3,5), (1,3), (0,0)} en R² ---")
A = sympy.Matrix([[3, 1, 0, 0],
                  [5, 3, 0, 0]])
sol = sympy.solve_linear_system(A, a1, a2, a3)
print("Solución sistema homogéneo:", sol, "  [a3 libre: multiplica al vector nulo]")
print("→", clasificar(sol, [a1, a2, a3]))

# --- (c) {(1,2,3,4), (1,0,0,0), (0,1,0,0)} en R⁴ ---
print("\n--- (c) {(1,2,3,4), (1,0,0,0), (0,1,0,0)} en R⁴ ---")
A = sympy.Matrix([[1, 1, 0, 0],
                  [2, 0, 1, 0],
                  [3, 0, 0, 0],
                  [4, 0, 0, 0]])
sol = sympy.solve_linear_system(A, a1, a2, a3)
print("Solución sistema homogéneo:", sol)
print("→", clasificar(sol, [a1, a2, a3]))

# --- (d) {(1,1,1), (2,2,2), (0,0,1)} en R³ ---
print("\n--- (d) {(1,1,1), (2,2,2), (0,0,1)} en R³ ---")
A = sympy.Matrix([[1, 2, 0, 0],
                  [1, 2, 0, 0],
                  [1, 2, 1, 0]])
sol = sympy.solve_linear_system(A, a1, a2, a3)
print("Solución sistema homogéneo:", sol)
print("Rango de la matriz:", np.linalg.matrix_rank(np.array([[1,2,0],[1,2,0],[1,2,1]], dtype=float)))
print("→", clasificar(sol, [a1, a2, a3]))

# --- (e) Matrices 2×2 como vectores en R⁴ (aplanadas por filas) ---
# M1=[[1,2],[3,4]] → [1,2,3,4]
# M2=[[1,0],[0,1]] → [1,0,0,1]
# M3=[[1,0],[0,0]] → [1,0,0,0]
print("\n--- (e) {M1, M2, M3} ∈ M₂ₓ₂  →  vectores en R⁴ ---")
print("  M1 = [1,2,3,4],  M2 = [1,0,0,1],  M3 = [1,0,0,0]")
A = sympy.Matrix([[1, 1, 1, 0],   # 1ª componente
                  [2, 0, 0, 0],   # 2ª componente
                  [3, 0, 0, 0],   # 3ª componente
                  [4, 1, 0, 0]])  # 4ª componente
sol = sympy.solve_linear_system(A, a1, a2, a3)
print("Solución sistema homogéneo:", sol)
print("→", clasificar(sol, [a1, a2, a3]))
```

## <center><font color=#3FA1BF>Ejercicio 2 — Autovalores y Autovectores</font></center>

Determinar los autovalores y autovectores de la matriz:

$$A = \begin{pmatrix} 1 & 4 & 2 \\ 4 & 1 & 2 \\ 2 & 2 & 3 \end{pmatrix}$$

Los **autovalores** $\lambda$ satisfacen el **polinomio característico** $\det(A - \lambda I) = 0$. Para cada $\lambda$, el **autovector** correspondiente satisface $(A - \lambda I)\mathbf{v} = \mathbf{0}$.

```python
# Ejercicio 2 — Autovalores y autovectores
import numpy as np
import sympy

A_np = np.array([[1, 4, 2],
                 [4, 1, 2],
                 [2, 2, 3]], dtype=float)

# Cálculo numérico con NumPy
eigenvals_np, eigenvects_np = np.linalg.eig(A_np)
print("Autovalores (NumPy):", eigenvals_np)
print("Autovectores (NumPy, columnas):\n", eigenvects_np)

# Cálculo simbólico exacto con SymPy
A_sp = sympy.Matrix([[1, 4, 2],
                     [4, 1, 2],
                     [2, 2, 3]])
print("\nAutovalores (SymPy):", A_sp.eigenvals())
print("\nAutovectores (SymPy):")
for lam, mult, vecs in A_sp.eigenvects():
    print(f"  λ = {lam}  (multiplicidad {mult}):  v = {vecs[0].T}")

# Verificación: A·v = λ·v
print("\nVerificación A·v = λ·v:")
for lam, mult, vecs in A_sp.eigenvects():
    v = vecs[0]
    igual = A_sp * v == lam * v
    print(f"  λ = {lam}: ¿A·v = λ·v? {igual}")
```

## <center><font color=#3FA1BF>Ejercicio 3 — Autovector conocido, encontrar constantes</font></center>

Para qué valores de $a$ y $c$ el vector $(8, 4)$ es un autovector de la matriz

$$A = \begin{pmatrix} a & 2 \\ c & 1 \end{pmatrix}$$

asociado al **autovalor** $\lambda = 3$.

**Planteamiento:** Si $(8,4)$ es autovector con autovalor $3$, entonces $A \cdot \mathbf{v} = 3\mathbf{v}$:

$$\begin{pmatrix} a & 2 \\ c & 1 \end{pmatrix}\begin{pmatrix}8\\4\end{pmatrix} = 3\begin{pmatrix}8\\4\end{pmatrix} \implies \begin{cases} 8a + 8 = 24 \\ 8c + 4 = 12 \end{cases}$$

```python
# Ejercicio 3 — Encontrar a y c tal que (8,4) es autovector con autovalor λ=3
import sympy

a, c = sympy.symbols('a c')
lam = sympy.Integer(3)

A = sympy.Matrix([[a, 2],
                  [c, 1]])
v = sympy.Matrix([8, 4])

# Condición: A·v = λ·v  →  cada componente de (A·v - λ·v) = 0
ecuaciones = [
    sympy.Eq((A * v)[0], lam * v[0]),   # fila 1: 8a + 8 = 24
    sympy.Eq((A * v)[1], lam * v[1]),   # fila 2: 8c + 4 = 12
]
print("Ecuaciones del sistema:")
for eq in ecuaciones:
    print(" ", eq)

sol = sympy.solve(ecuaciones, [a, c])
print("\nSolución:", sol)

# Verificación
A_sol = A.subs(sol)
print("\nMatriz A con los valores encontrados:")
print(A_sol)
print("A·v =", (A_sol * v).T)
print("λ·v =", (lam * v).T)
print("¿A·v = λ·v?", A_sol * v == lam * v)
```

## <center><font color=#3FA1BF>Ejercicio 4 — Tres autovectores dados, encontrar constantes</font></center>

Determine los valores de $a, p, c, b, q, r$ si $(1,1,0)$, $(-1,0,2)$ y $(0,1,-1)$ son **vectores propios** de la matriz:

$$A = \begin{pmatrix} a & 1 & p \\ b & 2 & q \\ c & -1 & r \end{pmatrix}$$

**Planteamiento:** Para cada vector $\mathbf{v}_i$ con autovalor $\lambda_i$ se cumple $A\mathbf{v}_i = \lambda_i \mathbf{v}_i$. Esto genera **9 ecuaciones** (3 filas $\times$ 3 vectores) en las 9 incógnitas $\{a, b, c, p, q, r, \lambda_1, \lambda_2, \lambda_3\}$.

```python
# Ejercicio 4 — Encontrar a,b,c,p,q,r a partir de tres autovectores
import sympy

a, b, c, p, q, r = sympy.symbols('a b c p q r')
lam1, lam2, lam3 = sympy.symbols('lambda1 lambda2 lambda3')

A = sympy.Matrix([[a,  1, p],
                  [b,  2, q],
                  [c, -1, r]])

v1 = sympy.Matrix([ 1,  1,  0])   # 1er vector propio
v2 = sympy.Matrix([-1,  0,  2])   # 2do vector propio
v3 = sympy.Matrix([ 0,  1, -1])   # 3er vector propio

# Cada condición A·vi = λi·vi genera 3 ecuaciones (una por fila)
ecuaciones = []
for vi, lami in [(v1, lam1), (v2, lam2), (v3, lam3)]:
    Av = A * vi
    lv = lami * vi
    for k in range(3):
        ecuaciones.append(sympy.Eq(Av[k], lv[k]))

print(f"Total de ecuaciones generadas: {len(ecuaciones)}")

# Resolver el sistema para todas las incógnitas
sol = sympy.solve(ecuaciones, [a, b, c, p, q, r, lam1, lam2, lam3])
print("\nSolución:")
for k, v in sol.items():
    print(f"  {k} = {v}")

# Construir A con los valores encontrados
A_sol = A.subs(sol)
print("\nMatriz A con los valores encontrados:")
print(A_sol)

# Verificación: A·vi = λi·vi para cada autovector
print("\nVerificación A·vi = λi·vi:")
for vi, lami, nombre in [(v1, lam1, "v1 = (1,1,0)"),
                          (v2, lam2, "v2 = (-1,0,2)"),
                          (v3, lam3, "v3 = (0,1,-1)")]:
    lam_val = sol[lami]
    residuo = sympy.simplify(A_sol * vi - lam_val * vi)
    igual = residuo == sympy.zeros(3, 1)
    print(f"  {nombre}: λ = {lam_val},  ¿A·v = λ·v? {igual}")
```
