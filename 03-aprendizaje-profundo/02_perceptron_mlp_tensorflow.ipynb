## 1. Implementación de un Perceptrón Simple en TensorFlow

**Objetivo**: Implementar un perceptrón simple para clasificar la función lógica AND.



Cree una red neuronal ANN para entrenar la función AND

```python
import numpy as np
import tensorflow as tf

# Datos de entrada (Función AND)
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [0], [0], [1]])  # Etiquetas
```

```python
# Se define el modelo. Se utiliza una capa Input para especificar la forma de entrada.
# En el ejemplo se tienen dos entradas y una salida.
model = tf.keras.Sequential([
    tf.keras.Input(shape=(2,)),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Se compila el modelo. Se usa binary_crossentropy dado que es una clasificacion binaria
# Cambiamos el optimizador a SGD con una tasa de aprendizaje específica para este problema pequeño.
model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.1),
                          loss='binary_crossentropy',
                          metrics=['accuracy'])

# Aumentamos las épocas de entrenamiento para permitir que el modelo converja mejor.
model.fit(X, y, epochs=1000, verbose=0)

# Evaluar el modelo
loss, accuracy = model.evaluate(X, y, verbose=0)
print(f"Precisión del modelo alternativo después del entrenamiento: {accuracy*100:.2f}%")

# Realizar predicciones
predictions = model.predict(X)
print("\nPredicciones del modelo alternativo (después de redondear):\n", np.round(predictions))
```

**Análisis**: El modelo inicialmente se compilaba usando el optimizador `Adam`, sin embargo, no se lograban métricas de accuracy superiores al 75%. Al cambiar el optimizador a un `SDG` con una tasa de aprendizaje de 0.1, el modelo logró representar la funcion AND apropiadamente.

## 3. Regularización y Dropout en Redes Neuronales

**Objetivo** Implementar técnicas de regularización (Dropout y L2) para evitar sobreentrenamiento.

 TAREA: Ajusta la cantidad de dropout y observa cómo afecta la precisión del modelo.


```python
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.utils import to_categorical

# Cargar los datos
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

# Normalizar los datos
X_train, X_test = X_train / 255.0, X_test / 255.0

# Convertir etiquetas a one-hot encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# Definir modelo con regularización y dropout
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation='relu', kernel_regularizer=tf.keras.regularizers.l2(0.1)),  # L2 REGULARIZATION
    tf.keras.layers.Dropout(0.2),  # DROPOUT
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

# Compilar el modelo
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Entrenar el modelo
model.fit(X_train, y_train, epochs=5, validation_data=(X_test, y_test))

# Evaluar el modelo
loss, acc = model.evaluate(X_test, y_test)
print(f'Precisión en test: {acc:.4f}')

```

## Corregir carga y tokenización del dataset IMDB

Se le solicitó a Gemini corregir la importación del dataset para mitigar el error presentado con el código original.

**Reasoning**:
The previous code failed because `info.features['text'].encoder.vocab_size` was `None`, indicating the `encoder` was not correctly loaded or accessible in the `info` object. To resolve this, I will modify the cell `CfG0CUM-Pgrx` to load the dataset without `as_supervised=True` initially, then create and adapt a `TextVectorization` layer to build the vocabulary from the training text. This `TextVectorization` layer will then be used to tokenize both training and test data. Finally, the model's `Embedding` layer will be updated with the correct vocabulary size and a user-defined `embedding_dim`, and the datasets will be batched accordingly, following all specified instructions.



```python
import tensorflow as tf
import tensorflow_datasets as tfds
from tensorflow.keras.layers import TextVectorization

# Cargar dataset IMDb de reseñas de películas
# Cargar el dataset. Se mapearán (text, label) luego.
(ds_train_original, ds_test_original), info = tfds.load(
    'imdb_reviews',
    split=('train','test'),
    # No as_supervised=True de inicialmente para poder procesar el diccionario
    with_info=True
)
#Extraer texto del dataset de entrenamiento para adaptación del vocabulario
train_texts_only = ds_train_original.map(lambda x: x['text'])

# Definir un tamaño máximo del vocabulario para la capa TextVectorization
# Una elección común para el análisis de sentimientos del IMDb es cerca de 2000 tokens
VOCAB_SIZE = 20000

# Crear la capa de TextVectorization
# output_sequence_length es importante para que la capa de Embedding layer
# conozca la forma de la entrada.
vectorize_layer = TextVectorization(
    max_tokens=VOCAB_SIZE,
    output_mode='int', # Output integer indices
    output_sequence_length=256 # Pad/truncate sequences to this length
)

# Adaptar la capa TextVectorization layer al texto de entrenamiento para construir el vocabulario
vectorize_layer.adapt(train_texts_only)

# Tomar el tamaño real del vocabulario de la capa adaptada
actual_vocab_size = len(vectorize_layer.get_vocabulary())
print(f"Vocabulary size after adaptation: {actual_vocab_size}")

# Definir el embedding_dim
embedding_dim = 64

# Preprocesar los datasets usando la capa TextVectorization
def preprocess_data(example):
    text = example['text']
    label = example['label']
    text = vectorize_layer(text)
    return text, label

ds_train = ds_train_original.map(preprocess_data)
ds_test = ds_test_original.map(preprocess_data)

# Construir modelo de red neuronal para texto
model = tf.keras.Sequential([
    # Update Embedding layer (step 5)
    tf.keras.layers.Embedding(actual_vocab_size, embedding_dim, mask_zero=True), # mask_zero=True allows for variable length input or padding
    tf.keras.layers.GlobalAveragePooling1D(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

# Compilar modelo
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Entrenar modelo
# Ajustar el tamaño de lote de ds_train and ds_test (step 6)
BATCH_SIZE = 64
ds_train = ds_train.shuffle(10000).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
ds_test = ds_test.batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

model.fit(ds_train, epochs=5, validation_data=ds_test)
```

* El conjunto de datos 'imdb_reviews' se cargó correctamente, y se empleó una capa TextVectorization con un VOCAB_SIZE especificado de 20.000 y una output_sequence_length de 256. Se confirmó que el tamaño real del vocabulario después de la adaptación era de 20.000.
* Ambos conjuntos de datos de entrenamiento y prueba se tokenizaron correctamente en secuencias de enteros utilizando la capa TextVectorization adaptada.
* El embedding_dim se definió explícitamente como una variable y se estableció en 64, lo que facilita su modificación.
* La capa tf.keras.layers.Embedding en el modelo de red neuronal se configuró correctamente con el tamaño de vocabulario determinado dinámicamente (20.000) y el embedding_dim (64), con mask_zero=True habilitado para manejar entradas de longitud variable o relleno.
