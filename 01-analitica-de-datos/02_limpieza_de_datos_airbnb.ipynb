## Punto 2: Identificación y Manejo de Valores Faltantes

**Instrucción:**

	1.	Identifica las columnas que contienen valores faltantes.
	2.	Decide qué hacer con los valores faltantes:
		•	Elimina filas o columnas que contengan valores faltantes.
		•	Rellena los valores faltantes con un valor adecuado (media, mediana, moda, etc.).

**Guía:**

	•	Utiliza .isnull().sum() para contar los valores faltantes en cada columna.
	•	Considera eliminar columnas si una gran parte de sus datos está faltando o rellenar los valores faltantes con la media o mediana, según corresponda.

**Código de Ejemplo:**
```python
# Decidir cómo manejar los valores faltantes
# Ejemplo: Eliminar columnas con demasiados valores faltantes
df = df.dropna(axis=1, thresh=int(0.5 * len(df)))  # Mantener columnas con al menos 50% de datos

# Ejemplo: Rellenar valores faltantes en una columna numérica con la media
df['columna_numerica'] = df['columna_numerica'].fillna(df['columna_numerica'].mean())
```

````python
# Identificar valores faltantes por columna
valores_faltantes = df.isnull().sum() # Completar
valores_faltantes
````

````python
# Tu código aquí

#------------------------------------------------------------------------------#

# Primero, se mejoran los nombres de las columnas para mejorar las presentación
df.columns = df.columns.str.replace(' ', '_')
df.columns = df.columns.str.lower()
print(df.columns)
df.info()
````

````python
# Inicialmente se pueden eliminar todas las filas que tengan todos sus valores
# faltantes
df = df.dropna(how='all')
df.info()
# En este caso no hubo cambios, lo que significa que todas las filas tienen al
# menos un dato

#------------------------------------------------------------------------------#

# Se revisan los datos diferentes en las columnas con valores faltantes.
# Si son muchos los valores únicos, se usa la propiedad `shape` para identificar
# el número de datos únicos.
# Luego, se propone una estrategia de relleno.

print('Name', df['name'].unique().shape)
print('Host identity verified', df['host_identity_verified'].unique())
print('Host name', df['host_name'].unique().shape)
print('neighbourhood_group', df['neighbourhood_group'].unique())
print('neighbourhood', df['neighbourhood'].unique().shape)
print('lat', df['lat'].unique().shape)
print("Country", df['country'].unique())
print("country code", df['country_code'].unique())
print("instant_bookable", df['instant_bookable'].unique())
print("cancellation_policy", df['cancellation_policy'].unique())
print('room type', df['room_type'].unique())
print('Construction year', df['construction_year'].unique())
print('price', df['price'].unique().shape)
print('service fee', df['service_fee'].unique().shape)
print('minimum nights', df['minimum_nights'].unique().shape)
print('number of reviews', df['number_of_reviews'].unique().shape)
print('last review', df['last_review'].unique().shape)
print('reviews per month', df['reviews_per_month'].unique().shape)
print('review rate number', df['review_rate_number'].unique())
print('calculated host listings count', df['calculated_host_listings_count'].unique())
print('availability 365', df['availability_365'].unique().shape)
print('house_rules', df['house_rules'].unique().shape)
print('license', df['license'].unique())
````

````python
# Se decide eliminar las filas cuyos valores nulos están en la columna 'price'
df = df.dropna(subset=['price'])

#------------------------------------------------------------------------------#

# También se decide eliminar las filas de los valores nulos de 'lat' y 'long'
df = df.dropna(subset=['lat', 'long'])

#------------------------------------------------------------------------------#

# Para facilitar las operaciones, se procede a cambiar los datos nulos en
# columnas que se pueden tratar y se castearán por los datos apropiados.

# Para las columnas que pueden ser enteras, se rellenan los datos con ceros
# y luego se hace un 'cast' a tipo int
columnas_como_enteros = ['construction_year','minimum_nights','number_of_reviews','calculated_host_listings_count','availability_365']
df[columnas_como_enteros] = df[columnas_como_enteros].fillna(0)
df[columnas_como_enteros] = df[columnas_como_enteros].astype(int)

#------------------------------------------------------------------------------#

# Columnas que se pueden tratar como flotantes.
# Las columnas 'service_fee' y 'price' deben tratarse como float, pero antes
# debe limpiarse su formato eliminando el signo '$' y las comas de los decimales
df[['price','service_fee']].head(20)
df['price']=df['price'].str.replace('$','').str.replace(',','').astype(float)
df['service_fee']=df['service_fee'].str.replace('$','').str.replace(',','').astype(float)

# Finalmente se rellenan los valores nulos de 'service_fee' por ceros, asumiendo
# que la cuota de servicio se incluye en el precio
df['service_fee'] = df['service_fee'].fillna(0)

#------------------------------------------------------------------------------#

# Para el 'review_rate_number', se llenan los valores nulos con la media
df['review_rate_number'] = df['review_rate_number'].fillna(df['review_rate_number'].mean())

#------------------------------------------------------------------------------#

# Para las 'reviews_per_month', se puede rellenar con la mediana

df['reviews_per_month']=df['reviews_per_month'].fillna(df['reviews_per_month'].median())

#------------------------------------------------------------------------------#

df.info()
df.isnull().sum()
````

````python
#La columna 'host_identity_verified' se rellena con el valor 'unconfirmed'
df['host_identity_verified']=df['host_identity_verified'].fillna('unconfirmed')

#------------------------------------------------------------------------------#

# Las columnas 'name', 'host_name', 'cancellation_policy' y 'house_rules' se
# rellenan con un nuevo valor 'unknown'
df[['name','host_name','cancellation_policy','house_rules']]=df[['name','host_name','cancellation_policy','house_rules']].fillna('unknown')

#------------------------------------------------------------------------------#

# Para las columnas 'country' y 'country_code' se pueden llenar sus valores
# nulos con el único valor de la columna, usando la moda
df['country'] = df['country'].fillna(df['country'].mode()[0])
df['country_code'] = df['country_code'].fillna(df['country_code'].mode()[0])
df[['country','country_code']]=df[['country','country_code']].astype(str)

#------------------------------------------------------------------------------#

# La columna 'instant_bookable' se rellena con 'false' los valores nulos
df['instant_bookable']=df['instant_bookable'].fillna('false')
df['instant_bookable']=df['instant_bookable'].astype(bool)

#------------------------------------------------------------------------------#

# La columna 'last_review' se rellena con un valor nominal, la fecha de creación
# de Airbnb 01/Ago/2008 y se transforma el tipo de dato a Datetime

df['last_review']=df['last_review'].fillna('01/08/2008')
df['last_review']=pd.to_datetime(df['last_review'],dayfirst=False)

#------------------------------------------------------------------------------#

df.info()
df.isnull().sum()
````

````python
# Para el caso de la columna 'neighbourhood_group', se puede crear un
# diccionario que asocie un 'neighbourhood' con su respectivo
# 'neighbourhood_group' y usarlo para rellenar los 'neighbourhood_group'
# faltantes:

# Al analizar los valores únicos, se observa que Manhattan y Brooklyn tienen
# errores ortográficos. Se corrigen los datos
df['neighbourhood_group'] = df['neighbourhood_group'].replace(['brookln', 'manhatan'], ['Brooklyn', 'Manhattan'])
df['neighbourhood_group'].unique()

# Filtro para identificar los 'neighbourhood' nulos y compararlos con su
# 'neighbourhood_group'. Total=15
filter_1 = df['neighbourhood'].isnull()
filter_1.sum()
df[filter_1].loc[:, ['neighbourhood_group', 'neighbourhood']]

# Filtro para identificar los 'neighbourhood_group' nulos y compararlos con su
# 'neighbourhood'. Total=28
filter_2 = df['neighbourhood_group'].isnull()
filter_2.sum()
df[filter_2].loc[:, ['neighbourhood_group', 'neighbourhood']]

# Filtro para identificar si hay pares 'neighbourhood' y 'neighbourhood_group'
# ambos vacíos. Total=0
filter_3 = df['neighbourhood'].isnull() & df['neighbourhood_group'].isnull()
df[filter_3].loc[:, ['neighbourhood_group', 'neighbourhood']]

# Se crea un diccionario usando como clave el 'neighbourhood' y como valor el
# 'neighbourhood_group',
# para poder determinar a qué grupo pertenece cada vecindario y así rellenar los
# faltantes
dict_neighbourhoods = dict(zip(df['neighbourhood'].dropna(), df['neighbourhood_group']))
dict_neighbourhoods.keys().__len__()

# Se reemplazan los 'neighbourhoud group' nulos con el valos correspondiente
# según el diccionario creado
df['neighbourhood_group'] = df['neighbourhood_group'].fillna(df['neighbourhood'].map(dict_neighbourhoods))

# Para el caso de la columna 'neighbourhood', como ya la columna 'neighbourhood
# group' tiene sus valores completos, se puede agrupar por 'neighbourhood_group'
# y reemplazar los valores nulos en la columna 'neighbourhood' por la moda
# de cada grupo de vecindario

df_grouped_by_neighbourhood_group = df.groupby('neighbourhood_group')['neighbourhood'].agg(lambda x: x.mode()[0])

# Se reemplazan los 'neighbourhoud' nulos con el valos correspondiente según el
# groupby creado
df['neighbourhood'] = df['neighbourhood'].fillna(df['neighbourhood_group'].map(df_grouped_by_neighbourhood_group))

#------------------------------------------------------------------------------#

df.info()
df.isnull().sum()
````

````python
# Finalmente la columna 'license' se descarta al tener muchos datos nulos
df = df.drop('license', axis=1)

#------------------------------------------------------------------------------#

df.info()
df.isnull().sum()
````

**Conceptos Importantes:**

	•	Evaluar la cantidad y distribución de valores faltantes para decidir el mejor enfoque.
	•	Rellenar o eliminar valores faltantes puede depender de la importancia de la columna y la cantidad de datos faltantes.

## Punto 5: Normalización, Estandarización y Tratamiento de Variables Categóricas

**Variables Numéricas y Categóricas**

	•	Variables Numéricas: Son aquellas que representan cantidades y pueden tomar valores numéricos, como la edad, precio, ingresos, etc.
	•	Variables Categóricas: Representan categorías o grupos, como el género, el tipo de propiedad, la ciudad, etc. No tienen un orden inherente.

**Selección de Columnas Numéricas y Categóricas**

Antes de aplicar la normalización, estandarización, y el tratamiento de variables categóricas, es importante identificar y seleccionar correctamente las columnas que corresponden a cada tipo de variable.

**Instrucción:**

	1.	Identifica y selecciona las columnas numéricas y categóricas del DataFrame.
	2.	Decide cuántas columnas numéricas vas a normalizar y cuántas vas a estandarizar.
	3.	Aplica normalización y estandarización a las columnas numéricas seleccionadas utilizando un ciclo for.
	4.	Realiza el tratamiento de las variables categóricas seleccionadas utilizando get_dummies.

**Guía:**

	•	Usa df.select_dtypes(include=[...]) para seleccionar columnas numéricas o categóricas.
	•	Normaliza los datos para escalar los valores entre 0 y 1.
	•	Estandariza los datos para que tengan una media de 0 y una desviación estándar de 1.
	•	Usa pd.get_dummies() para crear variables dummy a partir de las columnas categóricas.

### Selección de Columnas Numéricas:
	•	df.select_dtypes(include=['int64', 'float64']).columns selecciona todas las columnas del DataFrame que tienen tipos de datos numéricos (int64 y float64).
	•	Estas columnas se utilizarán para la normalización y estandarización.

````python
# Selección de columnas numéricas
columnas_numericas = df.select_dtypes(include=['int64', 'float64']).columns
print("Columnas Numéricas:\n", columnas_numericas)
````

### Selección de Columnas Categóricas:
	•	df.select_dtypes(include=['object']).columns selecciona todas las columnas que tienen datos categóricos (de tipo object), como texto o categorías.
	•	Estas columnas se convertirán en variables dummy.

````python
# Selección de columnas categóricas
columnas_categoricas = df.select_dtypes(include=['object']).columns
print("Columnas Categóricas:\n", columnas_categoricas)
````

### Normalización de Columnas Numéricas:
	•	El ciclo for recorre las columnas seleccionadas para normalizar y aplica la fórmula de normalización (df[col] - df[col].min()) / (df[col].max() - df[col].min()).
	•	El resultado es una nueva columna con el sufijo _normalizada que contiene los valores normalizados.

````python
# Normalización de columnas numéricas seleccionadas
#columnas_a_normalizar = columnas_numericas[:2]  # Ejemplo: selecciona las dos primeras columnas numéricas

#------------------------------------------------------------------------------#

# Se escogen las columnas a normalizar directamente por su nombre
columnas_a_normalizar = ['price','service_fee','review_rate_number']

for col in columnas_a_normalizar:
    df[col + '_normalizada'] = (df[col] - df[col].min())/(df[col].max() - df[col].min()) # Completar con la fórmula de la normalización
    print(f"Columna normalizada: {col}")

df.head()
````

### Estandarización de Columnas Numéricas:
	•	Otro ciclo for recorre las columnas seleccionadas para estandarizar y aplica la fórmula de estandarización (df[col] - df[col].mean()) / df[col].std().
	•	El resultado es una nueva columna con el sufijo _estandarizada que contiene los valores estandarizados.

````python
# Estandarización de columnas numéricas seleccionadas
# columnas_a_estandarizar = columnas_numericas[2:4]  # Ejemplo: selecciona la tercera y cuarta columna numérica

#------------------------------------------------------------------------------#

# Se escogen las columnas a estandarizar directamente por su nombre
columnas_a_estandarizar = ['price','service_fee','review_rate_number']

for col in columnas_a_estandarizar:
    df[col + '_estandarizada'] = (df[col] - df[col].mean()) / df[col].std() # Completar con la fórmula de la estandarización
    print(f"Columna estandarizada: {col}")

df.head()
````

## Tratamiento de Variables Categóricas:
	•	pd.get_dummies(df, columns=columnas_categoricas) convierte todas las columnas categóricas seleccionadas en variables dummy. Esto crea una columna binaria (0 o 1) para cada categoría presente en las columnas originales.

````python
# Tratamiento de variables categóricas: Convertir las columnas categóricas a variables dummy

# Se escogen las columnas 'neighbourhood_group' y 'room_type' para ser columnas
# categóricas
columnas_categoricas = ['neighbourhood_group','room_type']

df_dummies = pd.get_dummies(df, columns=columnas_categoricas, dtype=int) # Completar con la función para convertir a variables dummy
print(f"Columnas categóricas convertidas a variables dummy:\n{df_dummies.head()}")

df_dummies.info()
df_dummies.head()
````

**Conceptos Importantes:**

	•	La normalización es útil cuando necesitas comparar datos que están en diferentes escalas.
	•	La estandarización es importante cuando los datos necesitan ser comparados de manera consistente, con una distribución normal.
	•	Las variables categóricas necesitan ser convertidas a un formato numérico para su uso en modelos de machine learning y otros análisis cuantitativos.
