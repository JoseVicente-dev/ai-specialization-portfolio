#Ciclos

Los ciclos o bucles permiten repetir un bloque de código varias veces. Python soporta principalmente dos tipos de bucles: **for** y **while**.

- **ciclo for**: Itera sobre una secuencia (como una lista) y ejecuta un bloque de código por cada elemento.
- **ciclo while**: Repite un bloque de código mientras una condición sea verdadera.

````python
#Imprime todos los elementos de la matriz. El ciclo determina el tipo adecuado de dato según el iterable
for fila in matriz:
  fila_actual = fila
  print(fila_actual)
  for columna in fila_actual:
    print(columna)
````

````python
#Ciclo while. Estos ciclos deben usarse con cuidado pues pueden declararse erróneamente y volverse infinitos
contador = 1
while contador <5:
  print(contador)
  contador += 1 #Actualiza el contador. Equivalente a contador = contador + 1
````

## Cracion de una DataFrame

Puedes crear un DataFrame a partir de un **diccionario de listas**, donde las *claves* del diccionario se convierten en los *nombres de las columnas* y las *listas* en los *valores de las filas*:

````python
#Crear un df a partir de un diccionario de listas
#NOTA: todas las listas deben tener el mismo tamaño o longitud

list_dict={
    'Nombre':['Luisa','Baron','Conan','David','Ellie'],
    'Edad':[21,22,23,2,1],
    'Ciudad':['Lima','Bogotá','Quito','Paisandú','Buenos Aires']
}

df=pd.DataFrame(list_dict)
print(df)
print('\n')
df.head() #Permite ver la información como una tabla
````

También se puede crear a partir de una *Lista de listas*, en donde se le especifica al df al crearlo cuál será el nombre de las columnas 'header'

````python
lists_list = [['Ana',23,'Madrid'],
              ['Luis',34,'Barcelona'],
              ['Maarta',29,'Valencia'],
              ['Pedro',42,'Sevilla']]
#Se especifica el nombre de las columnas al crear el df
#-> pd.DataFrame(lista_de_listas,columns=['columna_1','columna_2','columna_N']) donde N es igual a la longitud de las sublistas
df = pd.DataFrame(lists_list, columns=['Nombre','Edad','Ciudad'])

print(df)
````

## Acceso a datos en un df: métodos **loc** e **iloc**

Pandas proporciona dos métodos principales para acceder a los datos en un DataFrame: 'loc' e 'iloc'.

- **loc**: Se utiliza para acceder a los datos por etiquetas (es decir, por nombres de columnas y filas). 'loc' es *inclusivo* tanto en las filas como en las columnas.

````python
print(df.loc[0])#Accede a la primera fila (index = 0)
````

````python
#Acceder a la columna nombre
print(df.loc[:,'Nombre'])#Accede a la columna 'Nombre'
````

````python
#Acceder al valor en una fila y columna específica ->valor único
print(df.loc[1,'Nombre'])#Accede al valor de la fila 1 y columna 'Nombre'
````

- **iloc**: Se utiliza para acceder a los datos por *índices numéricos*. Funciona de manera similar a cómo lo hacen los índices en listas o arrays en Python.

````python
print(df.iloc[0])#Accede a la primera fila (index = 0)
````

````python
#Acceder a la primera columna. En lugar de usar el nombre, se usa el index equivalente
print(df.iloc[:,0])#Accede a la primera columna (index = 0)
````

````python
#Acceder a una fila y columan específicas
print(df.iloc[1,0])#Accede al valor de la fila 1 y columna 0
````

## Limpieza de Datos

La limpieza de datos es un paso *crucial* en el proceso de análisis de datos. Antes de realizar cualquier análisis o modelado, es fundamental asegurarse de que los datos estén *limpios*, *consistentes* y en un f*ormato adecuado*.

¿Por qué es importante la limpieza de datos?

Los datos en bruto suelen estar llenos de errores, valores faltantes, duplicados y otros problemas que pueden distorsionar los resultados del análisis:

- **Eliminar errores e inconsistencias** -> Datos precisos y coherentes
- **Errores faltantes** -> cómo tratar los datos incompletos
- **Estandarizar formatos**
- **Preparar los datos** -> Listos para análisis estadísticos y predictivos

### Valores faltantes

Son aquellos datos que no están presentes en un conjunto de datos
- Error en la recolección
- Pérdida de datos
- Entradas no aplicables

Pueden afectar los resultados si no se manejan adecuadamente.

#### Identificar valores faltantes

**Metodo `info()`**: Retorna información sobre el DataFrame donde se puede identificar de forma exploratoria si existen o no valores faltantes o nulos, aunque sin saber la fila exacta:

El siguiente ejemplo es el output al usar el método `df.info()`:

```
<class 'pandas.core.frame.DataFrame'>
RangeIndex: 5 entries, 0 to 4
Data columns (total 3 columns):
 #   Column  Non-Null Count  Dtype  
---  ------  --------------  -----  
 0   Nombre  4 non-null      object
 1   Edad    4 non-null      float64
 2   Ciudad  4 non-null      object
dtypes: float64(1), object(2)
memory usage: 252.0+ bytes
```

- Se indica el rango de índices `RangeIndex`: en este caso son 5 entradas, que van de 0 a 4
- Las columnas totales `Data columns`: 3 en el ejemplo
- La tabla que sigue muestra los nombres de las columnas en `Column` , el contedo de no nulos `Non-Null Count` y el tipo de dato en cada columna `Dtype`
  - En este caso, el conteo de no nulos es de cuatro en cada columna, indicando que al menos una fila de cada columna tiene un dato nulo o `null`
- Finalmente un resumen de los tipos de dato `dtypes:` y el uso de memoria `memory usage`

````python
# Crear un DataFrame a partir de un diccionario de listas
# con valores faltantes

# Importar la librería pandas

import pandas as pd

# Crear un diccionario con listas de datos
data = {
'Nombre': ['Ana', 'Luis', 'Marta', 'Pedro', None],
'Edad': [23, 34, None, 42, 25],
'Ciudad': ['Madrid', 'Barcelona', None, 'Sevilla', 'Valencia']
}

# Crear un DataFrame a partir del diccionario
df = pd.DataFrame(data)

# Mostrar la info resumida del DataFrame
df.info()
````

### Datos duplicados: detección y manejo

Los **datos duplicados** son aquellas *filas* en un DataFrame que contienen los mismos valores en todas las columnas o en columnas específicas.
- Los duplicados pueden distorsionar el análisis, inflando artificialmente ciertos resultados

**Método `duplicated()`**: devuelve una *Serie* booleana que indica si una *fila* es un duplicado de una *fila anterior*

### Normalización y Estandarización de Datos

La normalización y estandarización son técnicas utilizadas para escalar los datos. La normalización convierte los datos a un rango específico (por ejemplo, 0 a 1), mientras que la estandarización transforma los datos para que tengan una media de 0 y una desviación estándar de 1.

#### Normalizacion

**Normalización min-max**: Escala los datos a un rango específico, como, restando el valor mínimo y dividiendo por el rango (max - min):
- $\frac{x-min}{max-min}$

Esto escala los valores para que estén en el rango de 0 a 1.

````python
# Normalizar la columna 'Edad'
df['Edad Normalizada'] = (df['Edad'] - df['Edad'].min()) / (df['Edad'].max() - df['Edad'].min())
# Mostrar el DataFrame con la columna normalizada
df
````

#### Estandarización

**Estandarización (Z-score)**: Transforma los datos restando la *media* y dividiendo por la *desviación estándar*, centrando los datos alrededor de 0 y con una desviación estándar de 1:

$z=\frac{x-\mu}{\sigma}$

````python
# Estandarizar la columna 'Edad'
df['Edad Estandarizada'] = (df['Edad'] - df['Edad'].mean()) / df['Edad'].std()
# Mostrar el DataFrame con la columna estandarizada
df
````

###### Funciones Lambda

**Función Lambda**:  pequeña función anónima que se define utilizando la palabra clave `lambda`.

A diferencia de las funciones normales que se definen usando `def`, las funciones lambda no tienen un nombre y se limitan a una sola expresión.

Estas funciones son útiles cuando necesitas una función simple y de corta duración, generalmente para ser utilizada como argumento en funciones de *orden superior* (como `map()`, `filter()`, y `apply()` en Pandas).

**Sintatix**
```python

 lambda args : expresión aplicada a los args
```
Donde:
- `lambda`: palabra reservada para declarar la función lambda
- args: son los parámetros de la función. Se separan por comas `,`
- expresión: es el cálculo que se aplica a los args y además es *retornado* por la función -> La expresión es única.

````python
# Ejemplo Básico: Función Lambda para retornar las suma de dos Números
suma = lambda x, y: x + y
print(suma(5, 3))  # Resultado: 8
````

### Tratamiento de Outliers o Valores atípicos

**Outliers**: Son valores que se desvían significativamente de otros valores en el conjunto de datos. Estos pueden distorsionar los resultados del análisis.

#### Identificación de outliers

Para esto se usa el concepto de *inter cuartiles* - **IQR** o **Interquartile Range** el cual mide el rango dentro del cual se encuentra la mitad central de los datos.

- **Cuartiles**: Un conjunto de datos se divide en cuatro partes iguales usando los cuartiles. El primer cuartil (Q1) marca el límite inferior del 25% de los datos, el segundo cuartil (Q2) es la mediana (50%), y el tercer cuartil (Q3) marca el límite inferior del 75% de los datos.

Es una medida de la dispersión de datos que indica la amplitud del intervalo que contiene la mitad central de una distribución.
- Se calcula restando el primer cuartil (**Q1**) del tercer cuartil (**Q3**).

Los valores que están fuera de *1.5 veces* el IQR por encima o por debajo del Q1 y Q3 se consideran **outliers**

insights_iqr_example.jpg

````python
# Detectar outliers en la columna 'Edad'
#Q1 = df['Edad'].quantile(0.40)
Q1 = df['Edad'].quantile(0.25)
#Q3 = df['Edad'].quantile(0.60)
Q3 = df['Edad'].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df['Edad'] < (Q1 - 1.5*IQR)) | (df['Edad'] > (Q3 + 1.5*IQR))]
print("Outliers:\n")
outliers #En este caso no se retorna nada en el DataFrame debido a que las edades no tienen outliers
````

#### Eliminación de Outliers

Una vez identificados, los outliers pueden ser eliminados o reemplazados. Para ellos usamos la virgulilla `~` o `not` que niega la condición, manteniendo lo valores que **NO** son outliers

````python
# Eliminar outliers en la columna 'Edad'
df_sin_outliers_1 = df[~((df['Edad']<(Q1-1.5*IQR)) | (df['Edad']>(Q3+1.5*IQR)))]

# NOTA: El siguiente método permite usar las palabras claves 'not' y 'or' para realizar el filtrado
df_sin_outliers_2 = df.query('not(`Edad`<(@Q1-1.5*@IQR) or `Edad`>(@Q3+1.5*@IQR))')

#Mostrar el df sin outliers
print("DataFrame sin outliers:\n")
print(df_sin_outliers_1)
print('\n')
print(df_sin_outliers_2)
````

#### Reemplazo de outliers

Se pueden reemplazar los Outliers por la mediana de la columna.

````python
# Cálculo de la mediana
mediana = df['Edad'].median()

# Reemplazo de los outliers por la mediana
df['Edad'] = df['Edad'].apply(lambda x: mediana if (x < (Q1 - 1.5*IQR)) or (x > (Q3 + 1.5*IQR)) else x)
df
````

#### Conversión a mayúsculas o minúsculas

- **`str.lower()`**: convierte todas las letras a minúsculas
- **`str.upper()`**: convierte todas las letras a mayúsculas
- **`str.title()`**: convierte la primera letra de cada palabra a mayúsculas y el resto las deja en minúsculas

````python
# Convertir texto en minúsculas
df['Nombre'] = df['Nombre'].str.lower()
print(df)
# Convertir texto en mayúsculas
df['Nombre'] = df['Nombre'].str.upper()
print(df)
# Convertir primera letra en mayúsculas y el resto en minúsculas
df['Nombre'] = df['Nombre'].str.title()
print(df)
````
