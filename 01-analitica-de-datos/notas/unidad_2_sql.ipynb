# Bases de Datos

## SQL

SQL es el estándar de facto para trabajar con bases de datos relacionales,
y es compatible con la mayoría de los sistemas de gestión de bases de datos (DBMS) como MySQL, PostgreSQL, SQL Server, Oracle y SQLite.

Tablas, las cuales están compuestas por filas y columnas. Cada tabla en una base de datos relacional puede estar relacionada con otras tablas mediante claves → Organizar datos en estructuras lógicas

### Tabla

**Tabla**: colección de datos organizados en filas y columnas. Cada tabla en una BD tiene un nombre *único*

- **Columna**: representa un campo o atributo dentro de una tabla: Nombre, edad, dirección

- **Fila**: representan **registros únicos** como un cliente, un producto, una transacción

## RDBMS - Relational Database Management System

**RDBMS**: software que permite Crear, Gestionar, Manipular bases de datos relacionales. Los más populares incluyen MySQL, PostgreSQL, Oracle, SQL Server y SQLite

  - **Integridad de datos**: datos almacenados precisos y coherentes

  - **Transaccional:** aseguran que las operaciones se realicen completamente o no se realicen en absoluto

  - **SQL**: Lenguaje de consulta estándar, aunque puede tener variantes según el sistema

## SQLite

**SQLite** es una base de datos integrada que no requiere un servidor separado para funcionar. Simple, eficiente y de fácil integración con aplicaciones más pequeñas. Ligera y de fácil configuración.

- **Portable** toda la BD es almacena en un solo archivo

- **Sin servidor**

- **Código abierto** y disponible de forma gratuita

## **Lenguaje de Definición de Datos (DDL - Data Definition Language)**

- Se utiliza para definir y estructurar la base de datos, como crear, modificar o eliminar tablas y otros objetos.
- Comandos comunes:
    - `CREATE` (crear tablas, índices, *bases de datos*, etc.)
    - `ALTER` (modificar estructuras existentes)
    - `DROP` (eliminar objetos)

#### Conceptos Clave:

- **Esquema de la tabla**: Define la estructura de la tabla y especifica los tipos de datos que pueden almacenar sus columnas.

- **Tipos de datos**: Es importante entender los diferentes tipos de datos en SQL, como `TEXT` para cadenas de texto, `INTEGER` para números enteros, y `REAL` para números decimales.

### Crear una tabla

```sql
CREATE TABLE empleados (
ID INTEGER PRIMARY KEY AUTOINCREMENT,    
Nombre TEXT NOT NULL,    
Edad INTEGER,    
Departamento TEXT,    
Salario REAL
);
```

## Lenguaje de Manipulación de Datos (DML - Data Manipulation Language)

- Permite manipular los datos almacenados en las tablas.
- Comandos comunes:
    - `SELECT` (consultar datos)
    - `INSERT` (insertar nuevos datos)
    - `UPDATE` (actualizar datos existentes)
    - `DELETE` (eliminar datos)

### Inserción de Datos - Create

Se usa el comando `INSERT INTO` para agregar nuevos registros a una tabla.

```sql
INSERT INTO nombre_de_la_tabla (columna1, columna2, columna3, ...)
VALUES (valor1, valor2, valor3, ...);
```

**Ejemplo**
```sql
INSERT INTO empleados (Nombre, Edad, Departamento, Salario)
VALUES ('Carlos Pérez', 34, 'Ventas', 45000);
```

**NOTA**: no es necesario especificar el campo `ID` si está configurado como `AUTOINCREMENT`, ya que se generará automáticamente.

#### Insertar múltiples registros

**Ejemplo**

```sql
INSERT INTO empleados (Nombre, Edad, Departamento, Salario)
VALUES
('Ana García', 28, 'Marketing', 40000),
('Luis Rodríguez', 41, 'Desarrollo', 50000),
('Marta Fernández', 36, 'Ventas', 47000);
```

### Consulta de datos - Read

Se utiliza el comando `SELECT` para consultar y recuperar datos de una o más tablas. Este comando es fundamental para recuperar datos.
- Se pueden seleccionar todas las tablas usando el sígno `*` o solo algunas columnas específicas.

**Ejemplo**

Seleccionar todas las columnas de la tabla
```sql
SELECT *
FROM empleados;
```
Seleccionar solo las columnas `Nombre`,`Edad` y `Salario`
```sql
SELECT Nombre, Edad, Salario
FROM empleados;
```

#### Filtrado de Datos usando WHERE

El comando `WHERE` se utiliza para filtrar los resultados de una consulta y devolver solo aquellos registros que cumplen con una condición específica.

**Sintaxis**

```sql
SELECT columna1, columna2, ...
FROM nombre_de_la_tabla
WHERE condición;
```

**Ejemplo**

Escoger los registros que cumplen con la condición de que el valor del campo `Departamento` sea 'Ventas'
```sql
SELECT Nombre, Salario
FROM empleados
WHERE Departamento = 'Ventas';
```


#### Uso de operadores lógicos AND, OR, NOT

Los operadores lógicos `AND`,`OR` y `NOT` permiten combinar múltiples condiciones en una consulta `WHERE`.

**Ejemplo**

Selecciona los registros que cumplan con la condición de que el campo `Departamento` sea *igual* a 'Ventas' y el `Salario` sea *mayor que* 45000
```sql
SELECT Nombre, Edad, Salario
FROM empleados
WHERE Departamento = 'Ventas' AND Salario > 45000;
```

### Actualización - Update y Eliminación de datos - Delete

Además de insertar y consultar datos, SQL permite modificar y eliminar registros existentes en una tabla.

#### Modificación de registros existentes - UPDATE

El comando `UPDATE` se utiliza para modificar los valores de uno o más registros en una tabla.

**Sintaxis**
```sql
UPDATE nombre_de_la_tabla
SET columna1 = valor1, columna2 = valor2, ...
WHERE condición;
```

**Ejemplo**
```sql
UPDATE empleados
SET Salario = 48000
WHERE Nombre = 'Carlos Pérez';
```


#### Eliminación de registros - DELETE

El comando `DELETE` se utiliza para eliminar uno o más registros de una tabla.

**Sintaxis**
```sql
DELETE FROM nombre_de_la_tabla
WHERE condición;
```

**Ejemplo**
```sql
DELETE FROM empleados
WHERE Edad < 25;
```

**ATENCIÓN**
- Es importante tener cuidado al usar `DELETE`. Si omites la cláusula `WHERE`, se eliminarán todos los *registros* de la tabla.

### Ordenamiento de Resultados - ORDER BY

`ORDER BY` se utiliza para ordenar los resultados de una consulta en orden ascendente `ASC`(por defecto) o descendente `DESC`

**Sintaxis**
```sql
SELECT columna1, columna2, ...
FROM nombre_de_la_tabla
ORDER BY columna1 [ASC|DESC];
```


### Agrupación de datos GROUP BY

`GROUP BY` se utiliza para agrupar **filas que tienen los mismos valores en columnas especificadas**  y luego realizar *cálculos agregados* en esos grupos, como sumar o contar registros.

**Sintaxis**
```sql
SELECT columna1, AGGREGATE_FUNCTION(columna2)
FROM nombre_de_la_tabla
GROUP BY columna1;
```

**Ejemplo**
```sql
SELECT Departamento, AVG(Salario)
FROM empleados
GROUP BY Departamento;
```

Esta consulta agrupa a los empleados por departamento y calcula el salario promedio (`AVG`) para cada grupo.

#### Uso de funciones agregadas

Estas *funciones* se usan junto a `GROUP BY` para realizar cálculos en grupos de datos:
- **`SUM`**: Suma todos los valores en un grupo
- **`COUNT`**: Cuenta el número de registros en un grupo
- **`AVG`**: Calcula la *media* de los valores en un grupo
- **`MIN`**: Encuentra el valor mínimo en un grupo
- **`MAX`**: Encuentra el valor máximo en un grupo

**Ejemplo**
```sql
SELECT Departamento, COUNT(*), SUM(Salario), MAX(Salario), MIN(Salario)
FROM empleados
GROUP BY Departamento;
```
Esta consulta *agrupa* los empleados por `Departamento` y calcula:
- El número total de empleados en cada departamento (`COUNT`).
- La suma total de salarios en cada departamento (`SUM`).
- El salario máximo y mínimo en cada departamento (`MAX` y `MIN`).


### Uniones básicas entre tablas

Las uniones permiten combinar datos de dos o más tablas basadas en una condición común. Esto es fundamental para trabajar con bases de datos relacionales, donde los datos se distribuyen en múltiples tablas.

#### INNER JOIN - Combinación de tablas relacionadas

`INNER JOIN` selecciona los registros que tienen coincidencias en ambas tablas involucradas en la unión. Es el tipo de unión más común y se utiliza cuando solo se necesitan los registros que tienen coincidencias en _**ambas**_ tablas.

**Sintaxis**
```sql
SELECT columnas
FROM tablaIzq
INNER JOIN tablaDer
ON tablaIzq.columna_comun = tablaDer.columna_comun;
```

**NOTA**: La columna común se iguala en ambas tablas con la sentencia `ON`

#### LEFT JOIN - Inclusión de filas sin correpondencia en la tabla derecha

`LEFT JOIN` retorna *todos* los registros de la **tabla izquierda** y los registros coincidentes de la **tabla derecha**:
  - Si no hay coincidencias, los resultados de la tabla derecha serán `NULL`

**Sintaxis**
```sql
SELECT columnas
FROM tablaIzq
LEFT JOIN tablaDer
ON tablaIzq.columna_comun = tablaDer.columna_comun;
```

#### RIGHT JOIN - Inclusión de filas sin correspondencia en la tabla izquierda

`RIGHT JOIN` retorna *todos* los registros de la **tabla izquierda** y los registros coincidentes de la **tabla derecha**:
  - Si no hay coincidencias, los resultados de la tabla izquierda serán `NULL`

**Sintaxis**
```sql
SELECT column_name(s)
FROM tableIzq
RIGHT JOIN tableDer
ON tableIzq.column_name = tableDer.column_name;
```

**NOTA**: In some databases `RIGHT JOIN` is called `RIGHT OUTER JOIN`.

#### FULL JOIN - Unión completa de las tablas

`FULL JOIN` retorna todos los datos de la **tabla izquierda** y todos los de la **tabla derecha**, incluso si no hay coincidencias.
  - `FULL JOIN` es lo mismo que `FULL OUTER JOIN`
  - Este comando puede retornar potencialmente resultados muy grandes

**Sintaxis**
```sql
SELECT column_name(s)
FROM table1
FULL OUTER JOIN table2
ON table1.column_name = table2.column_name
WHERE condition;
```

#### SELF JOIN - Unión de una tabla con sí misma

The self join, as its name implies, **joins** a table to *itself*.

To use a self join, the table must contain a column (call it `ColumnX`) that acts as the _**primary key**_ and a different column (call it `ColumnY`) that stores values that can be matched up with the values in `ColumnX`. The values of `ColumnX` and `ColumnY` *do not have to be the same for any given row*, and the value in `ColumnY` may even be `NULL`.

**Sintaxis**
```sql
SELECT TablaAlias1.columna1 AS ColumnaAlias1, TablaAlias2.columna2 AS ColumnaAlias2,...
FROM Tabla TablaAlias1
JOIN Tabla TablaAlias2
ON TablaAlias1.columnX = TablaAlias2.columnY
WHERE Condiciones...

```

**NOTA**:
  - When referring to the same table more than once in an SQL query, we need a way to distinguish each reference from the others. For this reason, it is important to use *aliases* to uniquely identify each reference of the same table in an SQL query. As a good practice, the aliases should indicate the role of the table for each specific reference in a query.
  - La sentencia anterior se puede especificar también sin el comando `JOIN` ... `ON` así:
  ```sql
  SELECT column_name(s)
  FROM table1 T1, table1 T2
  WHERE condition;
  ```


**Ejemplo**

La siguientes sentencias son _**equivalentes**_ y relacionan los `Customers` que son de la misma ciudad:

```sql
SELECT A.CustomerName AS CustomerName1, B.CustomerName AS CustomerName2, A.City
FROM Customers A
INNER JOIN Customers B
ON A.City = B.City
WHERE A.CustomerID <> B.CustomerID
ORDER BY A.City;
```
```sql
SELECT A.CustomerName AS CustomerName1, B.CustomerName AS CustomerName2, A.City
FROM Customers A, Customers B
WHERE A.CustomerID <> B.CustomerID
AND A.City = B.City
ORDER BY A.City;
```

### Subconsultas Básicas

Una **subconsulta** es una *consulta anidada* dentro de otra consulta.

Las subconsultas permiten realizar operaciones más complejas al permitirte utilizar el resultado de una consulta como entrada para otra.

Las subconsultas se pueden utilizar en varias partes de una consulta SQL, como en las cláusulas `SELECT`, `FROM`, y `WHERE`.

  - A menudo se usan para *comparar valores* o para *generar listas de valores* que se utilizan en la consulta principal.

#### Subconsulta en WHERE

```sql
SELECT Nombre, Salario
FROM empleados
WHERE Salario > (SELECT AVG(Salario) FROM empleados);
```
Esta consulta selecciona los nombres y salarios de todos los empleados cuyo salario es mayor que el salario promedio de todos los empleados.

>La subconsulta ```(SELECT AVG(Salario) FROM empleados)``` calcula el salario promedio, y este resultado se utiliza para filtrar los registros en la consulta principal.

#### Subconsulta en FROM

```sql
SELECT Departamento, MAX(Salario)
FROM (SELECT Departamento, Salario FROM empleados) GROUP BY Departamento;
```

Aquí, la subconsulta selecciona los departamentos y salarios de los empleados, y la consulta principal luego agrupa esos resultados por departamento y calcula el salario máximo en cada uno.

**NOTA**:
- Cabe notar que la subconsulta es una *tabla*, por tanto es perfectamente utilizable con el comando en el `FROM`

## Conexión Python con SQLite

**Python** incluye un módulo llamado **sqlite3** que permite interactuar con bases de datos SQLite de manera fácil y eficiente. Este módulo proporciona una interfaz para ejecutar comandos SQL directamente desde un script de Python.
  - No necesitas instalar sqlite3 ya que viene incluido en la instalación estándar de Python.

### Conexión a BD SQLite desde Python

**Establecer la conexión a la BD**: se hace mediante el comando `sqlite3.connect()`

**Ejemplo**

````python
import sqlite3

# Conectar a la base de datos (se creará si no existe)
conexion = sqlite3.connect('empresa.db')

# Crear un cursor para interactuar con la base de datos
cursor = conexion.cursor()
````

  - `sqlite3.connect('empresa.db')` conecta a la base de datos **empresa.db**. Si el archivo no existe, SQLite lo crea automáticamente.

  - `cursor = conexion.cursor()` crea un cursor que permite ejecutar comandos SQL a través de la conexión.


### Ejecución de comandos SQL desde Python

Una vez que tengas una **conexión** y un **cursor**, puedes ejecutar comandos SQL utilizando el método `execute()` del cursor.

**Ejemplo**

````python
# Crear una tabla desde Python
cursor.execute('''
CREATE TABLE IF NOT EXISTS empleados (
ID INTEGER PRIMARY KEY AUTOINCREMENT,
Nombre TEXT NOT NULL,
Edad INTEGER,
Departamento TEXT,
Salario REAL
)
''')

# Confirmar los cambios en la base de datos
conexion.commit()

# Podemos verificar que una nueva tabla ha sido creada consultando la tabla
# sqlite_master incorporada en SQLite, la cual ahora debería contener una
# entrada para la tabla movie
response = cursor.execute('SELECT name FROM sqlite_master')
response.fetchone()
````

  - `cursor.execute(…)` ejecuta el comando SQL para crear la tabla empleados.
  - `conexion.commit()` confirma los cambios en la base de datos, asegurando que la tabla se cree de manera permanente.

### Inserción de Datos desde Scripts Python

Puedes insertar datos en una tabla de la misma manera que lo harías en una sesión SQL normal, pero utilizando el método `execute()` desde Python.

**Ejemplo**


````python
# Insertar un nuevo registro en la tabla empleados


cursor.execute('''
INSERT INTO empleados (Nombre, Edad, Departamento, Salario)
VALUES (?, ?, ?, ?)
''', ('Carlos Pérez', 34, 'Ventas', 45000))

# Confirmar los cambios
conexion.commit()

response = cursor.execute('SELECT * from empleados')
response.fetchall()
````

   - Se utiliza un `?` como marcador de posición para cada valor que se insertará en la tabla, y luego se proporciona una *tupla* con los valores correspondientes.
    - SIempre use marcadores de posición en lugar de string formatting para unir valores Python a sentencias SQL, y así evitará SQL injection attacks

````python
# Insertar un nuevo registro en la tabla empleados

data = [
    ('Carlos Pérez', 34, 'Ventas', 45000),
    ('Jose Lopez', 33, 'Ventas', 40000),
    ('Ana Suarez', 32, 'Ventas', 50000),
    ('Carlos Yepes', 31, 'Ventas', 10000),
]

cursor.executemany('''
INSERT INTO empleados (Nombre, Edad, Departamento, Salario)
VALUES (?, ?, ?, ?)
''', data)

# Confirmar los cambios
conexion.commit()

response = cursor.execute('SELECT * from empleados')
response.fetchall()
````

### Actualización de registros desde Python

Además de insertar, también puedes actualizar y eliminar registros usando `UPDATE` y `DELETE` a través de Python.

**Ejemplo**

````python
# Actualizar el salario de un empleado
cursor.execute('''
UPDATE empleados
SET Salario = ?
WHERE Nombre = ?
''', (48000, 'Carlos Pérez'))

# Confirmar los cambios
conexion.commit()

response = cursor.execute('SELECT * from empleados')
response.fetchall()
````

### Eliminación de registros en Python

````python
# Eliminar un empleado de la tabla
cursor.execute('''
DELETE FROM empleados
WHERE Nombre = ?
''', ('Carlos Pérez',))

# Confirmar los cambios
conexion.commit()

response = cursor.execute('SELECT * from empleados')
response.fetchall()
````

- La estructura de los comandos `UPDATE` y `DELETE` sigue la misma lógica que `INSERT`, utilizando marcadores de posición para los valores.

### Procesamiento de consultas y resultados en Python

Después de ejecutar una consulta `SELECT`, puedes procesar los resultados utilizando métodos como `fetchone()`, `fetchall()`, o un **loop**.

**Ejemplo**

````python
# Recuperar todos los empleados del departamento de Ventas
cursor.execute('''
SELECT Nombre, Salario
FROM empleados
WHERE Departamento = ?
''',('Ventas',))

# Obtener todos los resultados
empleados_ventas = cursor.fetchall()

# Procesar los resultados
for empleado in empleados_ventas:
    print(f'Nombre: {empleado[0]}, Salario: {empleado[1]}')
````

**NOTA**
  - Es importante notar que al pasar la *tupla* como parámetro en la consulta, se ponga una coma (`,`) adicional al final de la tupla para no generar un error `ProgrammingError` en el código. Ej.: `('Ventas',)`

## Uso de SQL para análisis de datos en proyectos reales con Python

La integración de SQL con Python es especialmente útil en proyectos de análisis de datos donde necesitas extraer, transformar y cargar (**ETL**) grandes volúmenes de datos.

**Ejemplos**
  - **Análisis de ventas**: Extraer datos de ventas desde una base de datos SQLite, procesarlos con Python para calcular estadísticas como el promedio de ventas, y luego generar informes o gráficos.
  - **Gestión de inventarios**: Usar SQL para mantener una base de datos de inventarios y Python para automatizar alertas cuando el stock de un producto es bajo.

### Integración de SQL en Pipelines de Datos con Python

En proyectos de ciencia de datos, a menudo necesitas integrar SQL en pipelines de datos, donde los datos se extraen de bases de datos, se transforman con Python, y se cargan en otro sistema para análisis posterior.

**Ejemplo Pipeline**

  1. **Extracción**: Utilizar SQL para extraer datos de varias tablas relacionadas.

  2. **Transformación**: Procesar y limpiar los datos utilizando pandas en Python.

  3. **Carga**: Exportar los datos procesados a otro sistema o a un archivo para análisis posterior.
