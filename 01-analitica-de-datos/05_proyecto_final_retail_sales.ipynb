## 1. Carga de un archivo .csv o .xlsx y Almacenamiento en una Base de Datos Relacional usando SQL

### Instrucciones:
1. **Carga el archivo .csv o .xlsx:** Puedes utilizar cualquier archivo de datos en formato .csv o .xlsx que desees analizar.
2. **Crea una conexión a una base de datos relacional usando SQLite:** Vamos a usar SQLite para este proyecto, ya que es fácil de configurar y utilizar.
3. **Almacena los datos en la base de datos:** Crea una tabla en la base de datos y almacena los datos del archivo en ella.

```python
import pandas as pd
import sqlite3

# Se escoge el dataset numero 4. Retail Sales Data

# Cargar el archivo .csv o .xlsx
# Puedes cambiar 'archivo.csv' por el nombre de tu archivo, o usar un archivo .xlsx
# NOTA: Se usa el parámetro encoding = 'latin-1' debido a la codificacion del dataset
df = pd.read_csv('/content/Sample - Superstore.csv', encoding='latin-1')  # O usa pd.read_excel('archivo.xlsx')

# Conectar a la base de datos SQLite (crea una nueva base de datos si no existe)
conn = sqlite3.connect('proyecto_final.db')
cursor = conn.cursor()

# Convertir los nombres de columnas a un formato SQL-friendly (sin espacios ni caracteres especiales)
df.columns = df.columns.str.replace(' ', '_').str.replace(r'\W', '')
df.columns = df.columns.str.replace('-', '_').str.replace(r'\W', '')

# Almacenar el dataframe en una tabla SQL (reemplaza 'nombre_tabla' por el nombre que desees)
df.to_sql('superstore', conn, if_exists='replace', index=False)

# Verificar que los datos se hayan guardado correctamente
cursor.execute("SELECT * FROM superstore LIMIT 5;")
print(cursor.fetchall())

# Cerrar la conexión
conn.close()
```

## 2. Realizar una Consulta con SQL y Python para Determinar la Cantidad de Datos por Cada Columna

### Instrucciones:
1. **Conecta de nuevo a la base de datos.**
2. **Realiza una consulta SQL** para contar la cantidad de datos (no nulos) en cada columna de la tabla almacenada.
3. **Muestra los resultados** en el notebook.

```python
#Se revisan las columnas de la base de datos
conn = sqlite3.connect('proyecto_final.db')
cursor = conn.cursor()

cursor.execute("PRAGMA table_info(superstore);")
column_names = [row[1] for row in cursor.fetchall()]

print(column_names)

conn.close()
```

```python
# Reconectar a la base de datos SQLite
conn = sqlite3.connect('proyecto_final.db')
cursor = conn.cursor()

# Consultar la cantidad de datos no nulos en cada columna
# Usando la variable column_names creada
for row in column_names:
  query = """
  SELECT
      COUNT({}) as cantidad_datos
  FROM
      superstore;
  """.format(row)
  resultados = pd.read_sql_query(query, conn)
  print(row, resultados)

# Cerrar la conexión
conn.close()
```

**RESULTADO**: Se observa que todas las columnas tienen el mismo numero de datos, sugiriendo que no hay datos vacíos en ninguna

## 5. Realizar la Limpieza de Datos sobre las Columnas Seleccionadas

### Instrucciones:
1. **Elimina los duplicados** y maneja los valores faltantes en las columnas seleccionadas.
2. **Realiza cualquier otra limpieza necesaria,** como la conversión de tipos de datos o la eliminación de valores atípicos.

```python
# Eliminar duplicados
df_limpio = df_limpio.drop_duplicates()

# Manejar valores faltantes (puedes usar imputación o eliminación de filas)
df_limpio = df_limpio.dropna()  # También puedes usar .fillna() para imputar

# Se convierten las fechas a tipo date
df_limpio['Order_Date'] = pd.to_datetime(df_limpio['Order_Date'])
df_limpio['Ship_Date'] = pd.to_datetime(df_limpio['Ship_Date'])

# Se tratan algunas columnas string
df_limpio['Segment'] = df_limpio['Segment'].str.lower()
df_limpio['Region'] = df_limpio['Region'].str.lower()
df_limpio['Category'] = df_limpio['Category'].str.lower()

# Verificar el resultado
print(df_limpio.info())
print(df_limpio.describe())
```

```python
# Se analizan las distribuciones de los numéricos para identificar outliers
import matplotlib.pyplot as plt

for columna in ['Sales', 'Profit', 'Discount', 'Quantity']:
    plt.figure(figsize=(6, 2))
    plt.boxplot(df_limpio[columna], vert=False)
    plt.title(f'Boxplot para {columna}')
    plt.show()
```

```python
# Se observa que se tiene una gran cantidad de outliers para Sales y Profit
# Se identifican los outliers

# Detectar outliers en la columna 'Sales'
q1_sales = df_limpio['Sales'].quantile(0.25)
q3_sales = df_limpio['Sales'].quantile(0.75)
IQR_sales = q3_sales - q1_sales

outliers_sales = df_limpio[(df_limpio['Sales'] < (q1_sales - 1.5*IQR_sales)) | (df['Sales'] > (q3_sales + 1.5*IQR_sales))]
print("Cantidad de outliers en sales: ", outliers_sales['Order_ID'].count())
print("Porcentaje de outliers en sales: ", outliers_sales['Order_ID'].count()/df_limpio['Order_ID'].count())


# Detectar outliers en la columna 'Profit'
q1_profit = df_limpio['Profit'].quantile(0.25)
q3_profit = df_limpio['Profit'].quantile(0.75)
IQR_Profit = q3_profit - q1_profit

outliers_Profit = df_limpio[(df_limpio['Profit'] < (q1_profit - 1.5*IQR_Profit)) | (df['Profit'] > (q3_profit + 1.5*IQR_Profit))]
print("Cantidad de outliers en Profit: ", outliers_Profit['Order_ID'].count())
print("Porcentaje de outliers en Profit: ", outliers_Profit['Order_ID'].count()/df_limpio['Order_ID'].count())
```

* Se observa que el porcentaje de outliers para *Sales* y *Profit* es del 11.7% y el 18.8% del dataframe limpio. Se eliminan estos datos, iniciando por la columna Sales dado que es un porcentaje menor

```python
df_limpio = df_limpio[~((df_limpio['Sales']<(q1_sales-1.5*IQR_sales)) | (df_limpio['Sales']>(q3_sales+1.5*IQR_sales)))]

df_limpio['Order_ID'].count()
```

```python
df_limpio = df_limpio[~((df_limpio['Profit']<(q1_profit-1.5*IQR_Profit)) | (df_limpio['Profit']>(q3_profit+1.5*IQR_Profit)))]
df_limpio['Order_ID'].count()
```

```python
# Se normaliza la columna Sales para facilitar su análisis
df_limpio['Sales_normalizada'] = (df_limpio['Sales'] - df_limpio['Sales'].min()) / (df_limpio['Sales'].max() - df_limpio['Sales'].min())
df_limpio.describe()
```

```python
# Se estandariza la columna Profit
df_limpio['Profit_estandarizado'] = (df_limpio['Profit'] - df_limpio['Profit'].mean()) / df_limpio['Profit'].std()
df_limpio.describe()
```
